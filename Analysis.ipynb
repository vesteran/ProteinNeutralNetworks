```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
```

```python
df_1lyz = pd.read_csv("./1LYZ_Results/results.csv",header=None)
df_1lyz.head()
```

```python
plt.plot(df_1lyz[2],df_1lyz[1])
plt.title("QMEAN for 1LYZ During Adaptive Walk")
plt.xlabel("QMEAN")
plt.ylabel("Number of Attempted Mutations")
plt.show()
```

```python
df_1UBQ = pd.read_csv("./1UBQ_Results/results.csv",header=None)
plt.plot(df_1UBQ[2],df_1UBQ[1])
plt.title("QMEAN for 1UBQ During Adaptive Walk")

plt.xlabel("QMEAN")
plt.ylabel("Number of Attempted Mutations")
plt.show()
```
